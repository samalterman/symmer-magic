# Benchmark: 13-qubit Haar Random State
Comparing symplectic + FWHT method (serial vs parallel)

```python
from symmer import QuantumState
from symmermagic.utils import stab_entropy_symp
import time
```

```python
s = QuantumState.haar_random(n_qubits=13)
print(f"State generated: {s.n_qubits} qubits, d = {2**s.n_qubits}")
```

```python
t0 = time.perf_counter()
v_serial = stab_entropy_symp(s)
t_serial = time.perf_counter() - t0
print(f"symp serial:       SRE = {v_serial:.6f}   time = {t_serial:.2f}s")
```

```python
t0 = time.perf_counter()
v_par = stab_entropy_symp(s, parallel=True, n_proc=4)
t_par = time.perf_counter() - t0
print(f"symp parallel (4): SRE = {v_par:.6f}   time = {t_par:.2f}s")
```

```python
print(f"Values match: {abs(v_serial - v_par) < 1e-6}")
print(f"Speedup: {t_serial / t_par:.1f}x")
```
