In this notebook, we demonstrate how to calculate the stabilizer Renyi entropy for a QuantumState object

```python
from symmer import QuantumState
from symmermagic.utils import stab_renyi_entropy
import time
```

First things first, we'll generate a random state

```python
random_state=QuantumState.haar_random(n_qubits=12)
```

Okay, now we'll calcualte the exact SRE for this state. For an $n$ qubit state, this checks all $4^n$ Pauli strings. Because we're looking at a 12 qubit state, we'll parallelize the computation over 5 threads to reduce computation time

```python
t1=time.perf_counter()
m2_true=stab_renyi_entropy(random_state, approach='exact',parallel=True,n_proc=4,order=2)
t2=time.perf_counter()
print(m2_true)
print(t2-t1)
```

This computation is highly efficient for sparse molecules, and can be performed on a decently good laptop for states as large as 14 or 15 qubits. For larger dense states, it will be more efficient to use a sampling algorithm based on the Metropolis-Hastings algorithm, demonstrated below.

```python
t1=time.time()
m2_sample=stab_renyi_entropy(random_state,approach='Metropolis',n_samples=1e6)
t2=time.time()
print(m2_sample)
print(t2-t1)
```
